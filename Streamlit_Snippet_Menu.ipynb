## Block A: [Required] Imports & page setup

**What it does:** Loads the libraries and sets the browser-tab title and a wide layout. `st.set_page_config` must be the **first** Streamlit command in the file.

| Needs blocks | Goes in |
|---|---|
| Nothing | SLOT 1 (replace the hello-world lines) |

**Customize it:** Change the title, the caption, or the page title in the browser tab.

````python
# COPY → starter_app.py   (Block A: Imports & page setup)
import pandas as pd
import plotly.express as px
import streamlit as st

st.set_page_config(page_title="Penguin Explorer", layout="wide")

st.title("Palmer Penguins Explorer")
st.caption("An interactive look at 344 penguins from three islands in Antarctica.")
````

---
## Block B: [Required] Load the data (cached)

**What it does:** Reads the CSV into a DataFrame called `df`. Streamlit re-runs your whole file every time someone clicks something, so `@st.cache_data` remembers the result and skips re-reading the file.

| Needs blocks | Goes in |
|---|---|
| A | SLOT 2 |

**Customize it:** Try removing `.dropna()` and see what breaks later. (Then put it back!)

````python
# COPY → starter_app.py   (Block B: Load the data (cached))
@st.cache_data
def load_data():
    df = pd.read_csv("data/penguins.csv")
    df = df.dropna()  # drop rows with missing measurements
    return df


df = load_data()
````

---
## Block C: [Required] Sidebar filters

**What it does:** Adds widgets to a sidebar and creates `filtered`, a smaller DataFrame with only the penguins that match. **Every block after this uses `filtered`**, so the whole app reacts to the filters.

| Needs blocks | Goes in |
|---|---|
| B | SLOT 3 |

**Customize it:** Add a filter for `sex` using the same `multiselect` pattern as island.

````python
# COPY → starter_app.py   (Block C: Sidebar filters)
st.sidebar.header("Filters")

species = st.sidebar.multiselect(
    "Species",
    options=sorted(df["species"].unique()),
    default=sorted(df["species"].unique()),
)
islands = st.sidebar.multiselect(
    "Island",
    options=sorted(df["island"].unique()),
    default=sorted(df["island"].unique()),
)
mass_min, mass_max = st.sidebar.slider(
    "Body mass (g)",
    min_value=int(df["body_mass_g"].min()),
    max_value=int(df["body_mass_g"].max()),
    value=(int(df["body_mass_g"].min()), int(df["body_mass_g"].max())),
    step=50,
)

filtered = df[
    df["species"].isin(species)
    & df["island"].isin(islands)
    & df["body_mass_g"].between(mass_min, mass_max)
]

if filtered.empty:
    st.warning("No penguins match those filters. Try widening them in the sidebar.")
    st.stop()
````

## Block K: [Extra] BONUS: Predict the species with machine learning

**What it does:** Trains a **Random Forest** model on the data, then lets the user describe a penguin with sliders and guesses its species. This is the portfolio wow factor: a real ML model inside a web app.

`@st.cache_resource` trains the model once instead of on every click.

| Needs blocks | Goes in |
|---|---|
| B | SLOT 6 (or inside the Predict tab) |

**Customize it:** Remove `body_mass_g` from the features and see if the model gets less confident.

````python
# COPY → starter_app.py   (Block K: BONUS: Predict the species with machine learning)
from sklearn.ensemble import RandomForestClassifier

@st.cache_resource
def train_model(data):
    features = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
    model = RandomForestClassifier(n_estimators=100, random_state=42)
    model.fit(data[features], data["species"])
    return model

model = train_model(df)

st.subheader("Which species is this penguin?")
st.write("Move the sliders to describe a penguin and the model guesses its species.")
p1, p2 = st.columns(2)
bill_len = p1.slider("Bill length (mm)", 30.0, 60.0, 45.0)
bill_dep = p1.slider("Bill depth (mm)", 13.0, 22.0, 17.0)
flipper = p2.slider("Flipper length (mm)", 170.0, 235.0, 200.0)
mass = p2.slider("Body mass (g)", 2700, 6300, 4200, step=50)

new_penguin = pd.DataFrame(
    [[bill_len, bill_dep, flipper, mass]],
    columns=["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"],
)
prediction = model.predict(new_penguin)[0]
confidence = model.predict_proba(new_penguin).max()

st.success(f"Prediction: **{prediction}** ({confidence:.0%} confident)")
````

### When something breaks
| You see | It usually means |
|---|---|
| `NameError: name 'filtered' is not defined` | You pasted a block above the block that creates that variable. Check the letter order. |
| `IndentationError` | A line inside `with …:` or `if …:` isn't indented the same as its neighbors. |
| `StreamlitDuplicateElementId` | Two widgets look identical. Add `key="something_unique"` to one. |
| `set_page_config() can only be called once…` | `st.set_page_config` must be the first `st.` command. Move Block A to the very top. |
| `FileNotFoundError: data/penguins.csv` | Run `streamlit run` from inside the `penguins_streamlit_app` folder. |
